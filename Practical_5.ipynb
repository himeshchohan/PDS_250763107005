# 5. Perform read write and append operations on text file, csv files, binary files using python.

# Text file , Read file, Write file, append file

```python
# WRITE: Overwrite or create file
with open("sample.txt", "w", encoding="utf-8") as file:
    file.write("Hello World!\n")
    file.writelines(["Line 2\n", "Line 3\n"])

# APPEND: Add content to the end
with open("sample.txt", "a", encoding="utf-8") as file:
    file.write("This line is appended.\n")

# READ: Read contents
with open("sample.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
```

# CSV file

```python
import csv

# WRITE: Write new rows
with open("data.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Role"])
    writer.writerow(["Alice", "Developer"])

# APPEND: Append rows without overwriting
with open("data.csv", "a", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Bob", "Designer"])

# READ: Extract row by row
with open("data.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

# binary file

```python
import pickle

# WRITE: Serialize and save a dictionary object
with open("data.bin", "wb") as file:
    pickle.dump({"id": 1, "name": "Alice"}, file)

# APPEND: Append a secondary object to the file
with open("data.bin", "ab") as file:
    pickle.dump({"id": 2, "name": "Bob"}, file)

# READ: Deserialize sequentially until reaching End of File (EOF)
with open("data.bin", "rb") as file:
    while True:
        try:
            data = pickle.load(file)
            print(data)
        except EOFError:
            break
```
