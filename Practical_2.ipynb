# List

```python
help(list)
```

```python
l1 = ["Hello", 1, 5, 7, 55, 60, "ABC"]
print(l1)
print(type(l1))

l1.append("pqr")
print(l1)

# Copy before clearing
l3 = l1.copy()
print("Copy:", l3)

print("Count of 5:", l1.count(5))

l1.extend(["Python", "Data Science", 5, 15])
print(l1)

print("Index of 5:", l1.index(5))

print("Element at index 4:", l1[4])

l1.insert(5, "Software")
print(l1)

removed = l1.pop()
print("Removed:", removed)
print(l1)

l1.remove("Data Science")
print(l1)

l1.reverse()
print(l1)

# Now clear the list
l1.clear()
print("After clear:", l1)

# Second list
l2 = [5, 80, 70, 17]
print(l2)

l2.sort(reverse=True)
print("Descending:", l2)

l2.sort(reverse=False)
print("Ascending:", l2)
```

# Set

```python
help(set)
```

```python
set1 = {1, 2, 3 , 4, 5, 6, 7 ,8}
print(set1)

set1.add(9)
print(set1)

set2 = set1.copy()
print(set1)
print(set2)

set1.clear()
print(set1)

set3 = {5, 8, 15, 9, 22}
print(set2)
print(set3)
print(set3.difference(set2))
print(set3)

print(set2.difference_update(set3))
print(set2)

set2.discard(5)
print(set2)

set2.intersection(set3)
print(set2)

print(set2.intersection_update(set3))
print(set3.isdisjoint(set2))
print(set3.issubset(set2))
print(set3.issuperset(set2))
#print(set3.pop(9))
print(set3.remove(9))
print(set3.symmetric_difference(set2))
print(set3.symmetric_difference_update(set2))
set2 = set2.union(set3)
print(set2)

# set2.update(15)
# print(set2)
```

# Dictionary Creation and Manipulation

```python
help(dict)
```

```python
student = {
    "Name": "Himesh",
    "Age": 20,
    "Course": "Python"
}

print("Original Dictionary:")
print(student)

# Add a new key-value pair
student["City"] = "Pune"
print("\nAfter Adding City:")
print(student)

# Update a value
student["Age"] = 21
print("\nAfter Updating Age:")
print(student)

# Remove an item
student.pop("Course")
print("\nAfter Removing Course:")
print(student)

# Display keys, values and items
print("\nKeys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())

# Copy dictionary
new_student = student.copy()
print("\nCopied Dictionary:")
print(new_student)
```

# String Creation and Built-in Methods

```python
help(str)
```

```python
text = "Hello Python"

print("\nOriginal String:")
print(text)

print("Upper Case:", text.upper())
print("Lower Case:", text.lower())
print("Title Case:", text.title())
print("Length:", len(text))
print("Replace:", text.replace("Python", "World"))
print("Split:", text.split())
print("Starts with Hello:", text.startswith("Hello"))
print("Ends with Python:", text.endswith("Python"))
print("Find 'Python':", text.find("Python"))
print("Count of 'o':", text.count("o"))
```
