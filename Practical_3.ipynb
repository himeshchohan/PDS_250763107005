# if

```python
age = int(input("Age: "))
if age > 18:
    print("Yes Eligible for Voating")
```

# elif

```python
result = int(input("Enter Your Results: "))
if result > 90:
    print("A Grade")
elif result > 80:
    print("B Grade")
elif result > 70:
    print("C Grade")
elif result > 60:
    print("D Grade")

else:
    print("You have not passed in Exam.")
```

# check if given number odd or even.

```python
num = int(input("Enter Your Number: "))
if num % 2 == 0:
    print("Number is Even")
else:
    print("Number is Odd")
```

# While

```python
i = 0
while i < 10:
    i += 1
    print(i)
```
