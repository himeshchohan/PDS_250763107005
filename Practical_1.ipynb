# 1. Write a python program to demonstate variable, data types(int, float, str, bool) and all operators (arithmetic, assignment, comparison, logiacal, membership, identity, bitwise)

# Variables and Data types

```python
print("Variables and Data Types")
a = 5
b = 3.14
c = "Hello"
d = True
print(a)
print(b)
print(c)
print(d)
```

# Arithmetic Operator

```python
print("\n Arithmetic Operators")
x = 5
y = 10
Add = x + y
Sub = x - y
Mul = x * y
Div = x / y

print("Addition is: ", Add)
print("Subtraction is: ", Sub)
print("Multiplication is: ", Mul)
print("Division is: ", Div)
```

# Assignment Operator

```python
print("\n Assignment Operators")
x = 10
x += 5
print(x)
x -= 2
print(x)
```

# Comparison Operator

```python
print("\n Comparison Operators")
print("a == 10", a == 10)
print("a != 10", a != 10)
print("a > 10", a > 10)
print("a < 10", a < 10)
```

# Logical Operator

```python
print("\n Logical Operators")
a = True
b = False
print("a and b", a and b)
print("a or b", a or b)
print("not a", not a)
```

# Membership Operator

```python
print("\n Membership Operators")
list = [1, 2, 3, 4, 5]
print("2 in list", 2 in list)
print("6 not in list", 6 not in list)
```

# Identity Operator

```python
print("\n Identity Operators")
a = 5
b = 5
print("a is b", a is b)
print("a is not b", a is not b)
```

# Bitwise Operator

```python
print("\n Bitwise Operators")
a = 5
b = 3
print("a & b", a & b)
print("a | b", a | b)
print("a ^ b", a ^ b)
print("~a", ~a)
print("a << 2", a << 2)
print("a >> 2", a >> 2)
```
