# Write a Python Program using user define function with Different Arguments and Demonstrate use of Build in Module line math, random, statistics.

```python
# Without Return type
def himesh():
    print("Hello Himesh")
himesh()

# no return type With Argument
def sum(a, b):
    print("Sum: ", a+b)
sum(5, 10)

# Return type with no argument
def get_number():
    return 10
num = get_number()
print(num)

# Return type with Argument
def mul(a, b):
    return a * b
print(mul(5, 8))
```

```python
# Swap Function
def fun(a, b):
    return b, a

a, b = fun(3, 4)
print(a, b)

```

```python
import math
a = 2.25

print("ceil: ", math.ceil(a))
print("Exp: ", math.exp(4))
print("Factorial: ", math.factorial(5))
print("Floor: ", math.floor(a))
print("fmod: ", math.fmod(5, 2))
print("frexp: ", math.frexp(2))
print("Power: ", math.pow(5,2))
```

# Random

```python
import random
help(random)
```

```python
import random
print(random.random())

print(random.randint(1, 10))

print(random.uniform(5.5, 10.5))

items = ["Apple", "Banana", "Cherry"]
print(random.choice(items))

print(random.sample(items, k=2))

numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
```

```python
import statistics

data = [1, 2, 2, 3, 4, 7, 9]

print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Variance:", statistics.variance(data))
print("StDev:", statistics.stdev(data))
```
